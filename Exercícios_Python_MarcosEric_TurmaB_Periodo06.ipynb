## 21. Busca Binária

```python
def busca_binaria(lista, elemento):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == elemento:
            return meio
        elif lista[meio] < elemento:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1

numeros = [1, 3, 5, 7, 9, 11, 13]
elemento = int(input("Digite o número que deseja buscar: "))

resultado = busca_binaria(numeros, elemento)

if resultado != -1:
    print("Elemento encontrado na posição:", resultado)
else:
    print("Elemento não encontrado.")
```

## 22. Criptografia de Cifra de César

```python
def cifra_cesar(mensagem, deslocamento):
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():
            inicio = ord('A') if caractere.isupper() else ord('a')
            novo_caractere = chr(
                (ord(caractere) - inicio + deslocamento) % 26 + inicio
            )
            resultado += novo_caractere
        else:
            resultado += caractere

    return resultado

mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

print("Mensagem criptografada:", cifra_cesar(mensagem, deslocamento))
```

## 24. Validação de CPF

```python
def validar_cpf(cpf):
    cpf = ''.join(c for c in cpf if c.isdigit())

    if len(cpf) != 11:
        return False

    if cpf == cpf[0] * 11:
        return False

    soma = 0
    for i in range(9):
        soma += int(cpf[i]) * (10 - i)

    resto = soma % 11
    digito1 = 0 if resto < 2 else 11 - resto

    if digito1 != int(cpf[9]):
        return False

    soma = 0
    for i in range(10):
        soma += int(cpf[i]) * (11 - i)

    resto = soma % 11
    digito2 = 0 if resto < 2 else 11 - resto

    return digito2 == int(cpf[10])

cpf = input("Digite o CPF: ")

if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

## 30. Jogo da Forca

```python
import random

palavras = ["python", "programacao", "computador", "desenvolvimento"]
palavra = random.choice(palavras)

letras_descobertas = ["_"] * len(palavra)
tentativas = 6
letras_usadas = []

while tentativas > 0 and "_" in letras_descobertas:
    print("\nPalavra:", " ".join(letras_descobertas))
    print("Tentativas restantes:", tentativas)

    letra = input("Digite uma letra: ").lower()

    if letra in letras_usadas:
        print("Você já tentou essa letra.")
        continue

    letras_usadas.append(letra)

    if letra in palavra:
        for i in range(len(palavra)):
            if palavra[i] == letra:
                letras_descobertas[i] = letra
        print("Acertou!")
    else:
        tentativas -= 1
        print("Errou!")

if "_" not in letras_descobertas:
    print("\nParabéns! Você descobriu a palavra:", palavra)
else:
    print("\nVocê perdeu!")
    print("A palavra era:", palavra)
```

## 31. Ordenação por Bolha

```python
def bubble_sort(lista):
    n = len(lista)

    for i in range(n):
        for j in range(0, n - i - 1):
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista

numeros = [5, 3, 8, 1, 2, 7]

print("Lista original:", numeros)
print("Lista ordenada:", bubble_sort(numeros))
```

## 32. Jogo da Velha

```python
tabuleiro = [" "] * 9

def mostrar_tabuleiro():
    print()
    print(f"{tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]}")
    print("--+---+--")
    print(f"{tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]}")
    print("--+---+--")
    print(f"{tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]}")
    print()

def verificar_vitoria(jogador):
    combinacoes = [
        (0, 1, 2), (3, 4, 5), (6, 7, 8),
        (0, 3, 6), (1, 4, 7), (2, 5, 8),
        (0, 4, 8), (2, 4, 6)
    ]

    for a, b, c in combinacoes:
        if tabuleiro[a] == jogador and tabuleiro[b] == jogador and tabuleiro[c] == jogador:
            return True

    return False

jogador = "X"

for rodada in range(9):
    mostrar_tabuleiro()

    posicao = int(input(f"Jogador {jogador}, escolha uma posição (1-9): ")) - 1

    if posicao < 0 or posicao > 8 or tabuleiro[posicao] != " ":
        print("Posição inválida!")
        continue

    tabuleiro[posicao] = jogador

    if verificar_vitoria(jogador):
        mostrar_tabuleiro()
        print(f"Jogador {jogador} venceu!")
        break

    jogador = "O" if jogador == "X" else "X"
else:
    mostrar_tabuleiro()
    print("Empate!")
```

## 40. Lista de Compras

```python
lista = []

while True:
    print("\n--- LISTA DE COMPRAS ---")
    print("1 - Adicionar item")
    print("2 - Remover item")
    print("3 - Listar itens")
    print("4 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        item = input("Digite o item: ")
        lista.append(item)
        print("Item adicionado!")

    elif opcao == "2":
        item = input("Digite o item que deseja remover: ")

        if item in lista:
            lista.remove(item)
            print("Item removido!")
        else:
            print("Item não encontrado.")

    elif opcao == "3":
        if len(lista) == 0:
            print("A lista está vazia.")
        else:
            print("\nItens:")
            for item in lista:
                print("-", item)

    elif opcao == "4":
        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")
```

## 46. Calcular Média e Desvio Padrão

```python
import math

numeros = [10, 8, 7, 9, 6]

media = sum(numeros) / len(numeros)

soma = 0
for numero in numeros:
    soma += (numero - media) ** 2

variancia = soma / len(numeros)
desvio_padrao = math.sqrt(variancia)

print("Média:", media)
print("Desvio padrão:", desvio_padrao)
```

## 48. Ordenação por Inserção

```python
def insertion_sort(lista):
    for i in range(1, len(lista)):
        chave = lista[i]
        j = i - 1

        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = chave

    return lista

numeros = [5, 2, 9, 1, 5, 6]

print("Lista original:", numeros)
print("Lista ordenada:", insertion_sort(numeros))
```
